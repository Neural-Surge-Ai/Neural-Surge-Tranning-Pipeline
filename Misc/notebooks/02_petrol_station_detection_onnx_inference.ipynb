```python
from huggingface_hub import snapshot_download

model_dir = snapshot_download(
    repo_id="coolroman/petrol-station-detect",
    local_dir="/kaggle/working/petrol-model"
)

print("Downloaded:", model_dir)
```

```python
import os

for file in os.listdir("/kaggle/working/petrol-model"):
    print(file)
```

```python
import onnxruntime as ort

model_path = "/kaggle/working/petrol-model/best_fp16.onnx"

session = ort.InferenceSession(
    model_path,
    providers=["CUDAExecutionProvider", "CPUExecutionProvider"]
)

print("Model loaded successfully!")

print("Input:", session.get_inputs()[0].shape)
print("Output:", session.get_outputs()[0].shape)
```
