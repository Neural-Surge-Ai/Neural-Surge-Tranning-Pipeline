```python
import cv2
import json
import numpy as np

from paddleocr import PaddleOCR, TextRecognition


class HighAccuracyImageOCR:

    def __init__(
        self,
        device="gpu",
        retry_threshold=0.94,
        min_detection_confidence=0.40,
    ):
        """
        High accuracy OCR specifically for document/invoice IMAGES.

        device:
            "gpu"
            "gpu:0"
            "cpu"

        retry_threshold:
            OCR lines below this confidence are automatically
            cropped, enhanced, and recognized again.
        """

        self.retry_threshold = retry_threshold
        self.min_detection_confidence = min_detection_confidence

        # ---------------------------------------------------
        # Full OCR pipeline
        # ---------------------------------------------------

        self.ocr = PaddleOCR(

            # Current high-accuracy general OCR generation
            ocr_version="PP-OCRv6",

            # Use medium models explicitly for accuracy
            text_detection_model_name="PP-OCRv6_medium_det",
            text_recognition_model_name="PP-OCRv6_medium_rec",

            # Useful for photographed/scanned invoices
            use_doc_orientation_classify=True,
            use_doc_unwarping=True,
            use_textline_orientation=True,

            # Word-level boxes if needed later
            return_word_box=True,

            device=device,
        )

        # ---------------------------------------------------
        # Recognition-only model
        # Used for automatic retry of difficult regions
        # ---------------------------------------------------

        self.recognizer = TextRecognition(
            model_name="PP-OCRv6_medium_rec",
            device=device,
        )

    # =======================================================
    # IMAGE QUALITY
    # =======================================================

    def prepare_image(self, image_path):
        """
        Loads image and enlarges small document images.

        We DON'T aggressively threshold the full invoice because
        that can destroy thin characters, decimals, $, commas etc.
        """

        image = cv2.imread(image_path)

        if image is None:
            raise ValueError(
                f"Unable to read image: {image_path}"
            )

        h, w = image.shape[:2]

        # Small input images hurt OCR significantly.
        # Bring the long side to a reasonable document resolution.
        long_side = max(h, w)

        if long_side < 2000:

            scale = 2000 / long_side

            image = cv2.resize(
                image,
                None,
                fx=scale,
                fy=scale,
                interpolation=cv2.INTER_CUBIC,
            )

        return image

    # =======================================================
    # CREATE MULTIPLE VERSIONS OF A DIFFICULT TEXT LINE
    # =======================================================

    def make_variants(self, crop):

        variants = []

        # -----------------------------------
        # 1. Original crop
        # -----------------------------------

        variants.append({
            "name": "original",
            "image": crop
        })

        # -----------------------------------
        # 2. 2x upscale
        # -----------------------------------

        upscale_2 = cv2.resize(
            crop,
            None,
            fx=2,
            fy=2,
            interpolation=cv2.INTER_CUBIC,
        )

        variants.append({
            "name": "upscale_2x",
            "image": upscale_2
        })

        # -----------------------------------
        # 3. 3x upscale
        # -----------------------------------

        upscale_3 = cv2.resize(
            crop,
            None,
            fx=3,
            fy=3,
            interpolation=cv2.INTER_CUBIC,
        )

        variants.append({
            "name": "upscale_3x",
            "image": upscale_3
        })

        # -----------------------------------
        # 4. CLAHE contrast enhancement
        # -----------------------------------

        gray = cv2.cvtColor(
            upscale_3,
            cv2.COLOR_BGR2GRAY,
        )

        clahe = cv2.createCLAHE(
            clipLimit=2.0,
            tileGridSize=(8, 8),
        )

        enhanced = clahe.apply(gray)

        enhanced_bgr = cv2.cvtColor(
            enhanced,
            cv2.COLOR_GRAY2BGR,
        )

        variants.append({
            "name": "clahe",
            "image": enhanced_bgr
        })

        # -----------------------------------
        # 5. Otsu threshold
        # -----------------------------------

        _, otsu = cv2.threshold(
            enhanced,
            0,
            255,
            cv2.THRESH_BINARY +
            cv2.THRESH_OTSU,
        )

        otsu_bgr = cv2.cvtColor(
            otsu,
            cv2.COLOR_GRAY2BGR,
        )

        variants.append({
            "name": "otsu",
            "image": otsu_bgr
        })

        # -----------------------------------
        # 6. Adaptive threshold
        # -----------------------------------

        adaptive = cv2.adaptiveThreshold(
            enhanced,
            255,
            cv2.ADAPTIVE_THRESH_GAUSSIAN_C,
            cv2.THRESH_BINARY,
            31,
            11,
        )

        adaptive_bgr = cv2.cvtColor(
            adaptive,
            cv2.COLOR_GRAY2BGR,
        )

        variants.append({
            "name": "adaptive",
            "image": adaptive_bgr
        })

        return variants

    # =======================================================
    # RECOGNITION ONLY
    # =======================================================

    def recognize_variant(self, image):

        output = self.recognizer.predict(
            input=image,
            batch_size=1,
        )

        candidates = []

        for result in output:

            data = result.json["res"]

            text = str(
                data.get("rec_text", "")
            ).strip()

            confidence = float(
                data.get("rec_score", 0.0)
            )

            if text:

                candidates.append({
                    "text": text,
                    "confidence": confidence,
                })

        if not candidates:
            return None

        return max(
            candidates,
            key=lambda x: x["confidence"]
        )

    # =======================================================
    # CONSENSUS
    # =======================================================

    def choose_best_candidate(self, candidates):
        """
        Multiple preprocessing variants may return:

            INV-100028
            INV-100028
            INV-100O28
            INV-100028

        Consensus is more reliable than simply trusting one OCR run.
        """

        grouped = {}

        for candidate in candidates:

            text = candidate["text"].strip()

            # Normalize only spacing.
            # Do NOT remove punctuation because it matters for invoices.
            normalized = " ".join(
                text.split()
            )

            if normalized not in grouped:

                grouped[normalized] = {
                    "text": text,
                    "confidences": [],
                    "sources": [],
                }

            grouped[normalized][
                "confidences"
            ].append(
                candidate["confidence"]
            )

            grouped[normalized][
                "sources"
            ].append(
                candidate["source"]
            )

        ranked = []

        for normalized, data in grouped.items():

            confidences = data["confidences"]

            max_conf = max(confidences)
            avg_conf = sum(confidences) / len(
                confidences
            )

            support = len(confidences)

            # Recognition confidence +
            # agreement between variants.
            consensus_score = (
                max_conf * 0.65
                + avg_conf * 0.30
                + min(support / 6, 1) * 0.05
            )

            ranked.append({
                "text": data["text"],
                "confidence": max_conf,
                "average_confidence": avg_conf,
                "support": support,
                "sources": data["sources"],
                "consensus_score": consensus_score,
            })

        ranked.sort(
            key=lambda x: x["consensus_score"],
            reverse=True,
        )

        return ranked[0]

    # =======================================================
    # AUTOMATIC LOW-CONFIDENCE RETRY
    # =======================================================

    def retry_region(self, crop):

        variants = self.make_variants(crop)

        candidates = []

        for variant in variants:

            result = self.recognize_variant(
                variant["image"]
            )

            if result is None:
                continue

            candidates.append({
                "text": result["text"],
                "confidence": result["confidence"],
                "source": variant["name"],
            })

        if not candidates:
            return None

        return self.choose_best_candidate(
            candidates
        )

    # =======================================================
    # MAIN EXTRACTION
    # =======================================================

    def extract(self, image_path):

        image = self.prepare_image(
            image_path
        )

        # -----------------------------------------------
        # First full document OCR
        # -----------------------------------------------

        results = self.ocr.predict(
            image,

            # Important:
            # keep everything so WE can decide
            # what needs retry.
            text_rec_score_thresh=0.0,

            text_det_box_thresh=
                self.min_detection_confidence,
        )

        final_elements = []

        for result in results:

            data = result.json["res"]

            texts = data.get(
                "rec_texts",
                []
            )

            scores = data.get(
                "rec_scores",
                []
            )

            boxes = data.get(
                "rec_boxes",
                []
            )

            polygons = data.get(
                "rec_polys",
                []
            )

            detection_scores = data.get(
                "dt_scores",
                []
            )

            # ----------------------------------------
            # IMPORTANT
            #
            # Paddle gives us the processed image.
            # Bounding coordinates correspond to this
            # document-processing result.
            # ----------------------------------------

            visual_data = result.img

            if "preprocessed_img" in visual_data:

                processed_pil = visual_data[
                    "preprocessed_img"
                ]

                processed_image = np.array(
                    processed_pil
                )

                # PIL -> OpenCV
                processed_image = cv2.cvtColor(
                    processed_image,
                    cv2.COLOR_RGB2BGR,
                )

            else:

                processed_image = image

            ph, pw = processed_image.shape[:2]

            # ----------------------------------------
            # Process every detected text region
            # ----------------------------------------

            for index, (
                text,
                confidence,
                box,
            ) in enumerate(
                zip(
                    texts,
                    scores,
                    boxes,
                )
            ):

                text = str(text).strip()

                if not text:
                    continue

                confidence = float(
                    confidence
                )

                x1, y1, x2, y2 = [
                    int(v)
                    for v in box
                ]

                # ----------------------------------
                # Initial result
                # ----------------------------------

                element = {

                    "text": text,

                    "confidence":
                        confidence,

                    "original_text":
                        text,

                    "original_confidence":
                        confidence,

                    "bbox": [
                        x1,
                        y1,
                        x2,
                        y2,
                    ],

                    "retried": False,
                }

                if index < len(polygons):

                    element["polygon"] = (
                        np.asarray(
                            polygons[index]
                        )
                        .astype(int)
                        .tolist()
                    )

                # ==================================
                # Automatic retry
                # ==================================

                if (
                    confidence
                    < self.retry_threshold
                ):

                    # Padding is very important.
                    # Recognition models dislike
                    # text cropped too tightly.

                    box_width = x2 - x1
                    box_height = y2 - y1

                    pad_x = max(
                        8,
                        int(
                            box_width * 0.06
                        )
                    )

                    pad_y = max(
                        5,
                        int(
                            box_height * 0.25
                        )
                    )

                    cx1 = max(
                        0,
                        x1 - pad_x
                    )

                    cy1 = max(
                        0,
                        y1 - pad_y
                    )

                    cx2 = min(
                        pw,
                        x2 + pad_x
                    )

                    cy2 = min(
                        ph,
                        y2 + pad_y
                    )

                    crop = processed_image[
                        cy1:cy2,
                        cx1:cx2
                    ]

                    if crop.size > 0:

                        retry = (
                            self.retry_region(
                                crop
                            )
                        )

                        if retry:

                            element[
                                "retried"
                            ] = True

                            element[
                                "retry_result"
                            ] = retry

                            # Only replace initial
                            # recognition if retry
                            # has stronger evidence.

                            if (
                                retry[
                                    "confidence"
                                ]
                                >
                                confidence
                            ):

                                element[
                                    "text"
                                ] = retry[
                                    "text"
                                ]

                                element[
                                    "confidence"
                                ] = retry[
                                    "confidence"
                                ]

                final_elements.append(
                    element
                )

        # ==========================================
        # Reading order
        # ==========================================

        final_elements.sort(
            key=lambda item: (
                item["bbox"][1],
                item["bbox"][0],
            )
        )

        # ==========================================
        # Plain text
        # ==========================================

        full_text = "\n".join(
            item["text"]
            for item
            in final_elements
        )

        return {
            "image": image_path,
            "text": full_text,
            "elements": final_elements,
        }


# =========================================================
# RUN
# =========================================================

if __name__ == "__main__":

    extractor = HighAccuracyImageOCR(
        device="gpu",
        retry_threshold=0.94,
    )

    result = extractor.extract(
        "/kaggle/input/datasets/ahmedshayan1/sample12/1131w-pyLVaYlAk1o.webp"
    )

    print(
        result["text"]
    )

    with open(
        "invoice_ocr.json",
        "w",
        encoding="utf-8",
    ) as f:

        json.dump(
            result,
            f,
            indent=2,
            ensure_ascii=False,
        )
```

```python
import cv2
import json
import numpy as np
from pathlib import Path

from paddleocr import PaddleOCR, TextRecognition


class HighAccuracyImageOCR:
    """
    High-accuracy OCR pipeline for document / invoice IMAGES.

    Pipeline
    --------
    Image
      -> conservative resize
      -> orientation correction
      -> document unwarping
      -> text line orientation
      -> PP-OCRv6 text detection
      -> PP-OCRv6 recognition
      -> low-confidence automatic crop
      -> multiple crop enhancement variants
      -> recognition retry
      -> consensus voting
      -> reading-order reconstruction
      -> JSON + optional visualization

    No manual bounding boxes are required.
    """

    def __init__(
        self,
        device="cpu",

        # OCR confidence below this triggers retry
        retry_threshold=0.94,

        # Lower detection threshold improves recall
        min_detection_confidence=0.40,

        # Initial image size
        min_long_side=2000,

        # CPU controls
        cpu_threads=4,

        # Save/debug behavior
        verbose=True,
    ):

        self.device = device
        self.retry_threshold = retry_threshold
        self.min_detection_confidence = min_detection_confidence
        self.min_long_side = min_long_side
        self.verbose = verbose

        if self.verbose:
            print("=" * 60)
            print("Initializing High Accuracy OCR")
            print("Device:", device)
            print("Retry threshold:", retry_threshold)
            print("Detection threshold:", min_detection_confidence)
            print("=" * 60)

        # ====================================================
        # COMPLETE OCR PIPELINE
        # ====================================================

        self.ocr = PaddleOCR(

            # Explicit PP-OCRv6 models.
            # We do NOT specify ocr_version because explicit
            # model names make ocr_version unnecessary.

            text_detection_model_name=
                "PP-OCRv6_medium_det",

            text_recognition_model_name=
                "PP-OCRv6_medium_rec",

            # Document correction
            use_doc_orientation_classify=True,
            use_doc_unwarping=True,
            use_textline_orientation=True,

            # Useful if word-level data is needed later
            return_word_box=True,

            # CPU
            device=device,

            # IMPORTANT:
            # Disable oneDNN because your current
            # Paddle 3.3.1 CPU build crashes there.
            enable_mkldnn=False,

            cpu_threads=cpu_threads,
        )

        # ====================================================
        # RECOGNITION-ONLY MODEL
        #
        # Used on difficult automatically cropped regions.
        # ====================================================

        self.recognizer = TextRecognition(

            model_name=
                "PP-OCRv6_medium_rec",

            device=device,

            enable_mkldnn=False,

            cpu_threads=cpu_threads,
        )

        if self.verbose:
            print("OCR models initialized successfully.")


    # ========================================================
    # LOAD + PREPARE IMAGE
    # ========================================================

    def prepare_image(self, image_path):
        """
        Conservatively increase resolution when input is small.

        We deliberately DO NOT:
          - threshold the entire document
          - sharpen aggressively
          - remove colors

        because doing so can damage:
          decimal points
          commas
          currency signs
          thin digits
          invoice IDs
        """

        image = cv2.imread(
            str(image_path),
            cv2.IMREAD_COLOR
        )

        if image is None:
            raise ValueError(
                f"Unable to read image: {image_path}"
            )

        original_h, original_w = image.shape[:2]

        long_side = max(
            original_h,
            original_w
        )

        scale = 1.0

        if long_side < self.min_long_side:

            scale = (
                self.min_long_side
                / float(long_side)
            )

            image = cv2.resize(
                image,
                None,
                fx=scale,
                fy=scale,
                interpolation=cv2.INTER_CUBIC,
            )

        if self.verbose:
            print()
            print("Original image:",
                  original_w,
                  "x",
                  original_h)

            print("OCR image:",
                  image.shape[1],
                  "x",
                  image.shape[0])

            print("Input scale:",
                  round(scale, 3))

        return image, {
            "original_width": original_w,
            "original_height": original_h,
            "input_scale": scale,
        }


    # ========================================================
    # PIL / NUMPY RESULT -> OPENCV
    # ========================================================

    def to_opencv(self, image):
        """
        Convert Paddle/PIL image result safely to OpenCV BGR.
        """

        array = np.asarray(image)

        # grayscale
        if len(array.shape) == 2:

            return cv2.cvtColor(
                array,
                cv2.COLOR_GRAY2BGR
            )

        # RGBA
        if array.shape[2] == 4:

            return cv2.cvtColor(
                array,
                cv2.COLOR_RGBA2BGR
            )

        # RGB -> BGR
        return cv2.cvtColor(
            array,
            cv2.COLOR_RGB2BGR
        )


    # ========================================================
    # AUTOMATIC TEXT CROP
    # ========================================================

    def crop_region(
        self,
        image,
        bbox,
        pad_x_ratio=0.06,
        pad_y_ratio=0.25,
    ):

        x1, y1, x2, y2 = bbox

        h, w = image.shape[:2]

        box_w = max(
            x2 - x1,
            1
        )

        box_h = max(
            y2 - y1,
            1
        )

        # Add contextual padding around detected line
        pad_x = max(
            8,
            int(box_w * pad_x_ratio)
        )

        pad_y = max(
            5,
            int(box_h * pad_y_ratio)
        )

        cx1 = max(
            0,
            x1 - pad_x
        )

        cy1 = max(
            0,
            y1 - pad_y
        )

        cx2 = min(
            w,
            x2 + pad_x
        )

        cy2 = min(
            h,
            y2 + pad_y
        )

        crop = image[
            cy1:cy2,
            cx1:cx2
        ]

        return crop


    # ========================================================
    # CREATE OCR VARIANTS
    # ========================================================

    def make_variants(self, crop):
        """
        Generate several versions of difficult text.

        Important:
        preprocessing is performed ONLY on uncertain text,
        not on the whole invoice.
        """

        variants = []

        if crop is None or crop.size == 0:
            return variants

        # ----------------------------------------------------
        # 1. Original
        # ----------------------------------------------------

        variants.append({
            "name": "original_crop",
            "image": crop
        })

        # ----------------------------------------------------
        # 2. 2x upscale
        # ----------------------------------------------------

        upscale_2 = cv2.resize(
            crop,
            None,
            fx=2.0,
            fy=2.0,
            interpolation=cv2.INTER_CUBIC,
        )

        variants.append({
            "name": "upscale_2x",
            "image": upscale_2
        })

        # ----------------------------------------------------
        # 3. 3x upscale
        # ----------------------------------------------------

        upscale_3 = cv2.resize(
            crop,
            None,
            fx=3.0,
            fy=3.0,
            interpolation=cv2.INTER_CUBIC,
        )

        variants.append({
            "name": "upscale_3x",
            "image": upscale_3
        })

        # ----------------------------------------------------
        # Convert to grayscale
        # ----------------------------------------------------

        gray = cv2.cvtColor(
            upscale_3,
            cv2.COLOR_BGR2GRAY,
        )

        # ----------------------------------------------------
        # 4. CLAHE
        # ----------------------------------------------------

        clahe = cv2.createCLAHE(
            clipLimit=2.0,
            tileGridSize=(8, 8),
        )

        enhanced = clahe.apply(
            gray
        )

        enhanced_bgr = cv2.cvtColor(
            enhanced,
            cv2.COLOR_GRAY2BGR,
        )

        variants.append({
            "name": "clahe",
            "image": enhanced_bgr
        })

        # ----------------------------------------------------
        # 5. OTSU
        # ----------------------------------------------------

        _, otsu = cv2.threshold(
            enhanced,
            0,
            255,
            cv2.THRESH_BINARY
            + cv2.THRESH_OTSU,
        )

        otsu_bgr = cv2.cvtColor(
            otsu,
            cv2.COLOR_GRAY2BGR,
        )

        variants.append({
            "name": "otsu",
            "image": otsu_bgr
        })

        # ----------------------------------------------------
        # 6. Adaptive threshold
        # ----------------------------------------------------

        adaptive = cv2.adaptiveThreshold(

            enhanced,

            255,

            cv2.ADAPTIVE_THRESH_GAUSSIAN_C,

            cv2.THRESH_BINARY,

            31,

            11,
        )

        adaptive_bgr = cv2.cvtColor(
            adaptive,
            cv2.COLOR_GRAY2BGR,
        )

        variants.append({
            "name": "adaptive",
            "image": adaptive_bgr
        })

        return variants


    # ========================================================
    # RECOGNIZE ONE TEXT CROP
    # ========================================================

    def recognize_variant(self, image):

        if image is None or image.size == 0:
            return None

        outputs = self.recognizer.predict(
            input=image,
            batch_size=1,
        )

        candidates = []

        for result in outputs:

            data = result.json.get(
                "res",
                {}
            )

            text = str(
                data.get(
                    "rec_text",
                    ""
                )
            ).strip()

            confidence = float(
                data.get(
                    "rec_score",
                    0.0
                )
            )

            if text:

                candidates.append({
                    "text": text,
                    "confidence": confidence,
                })

        if not candidates:
            return None

        return max(
            candidates,
            key=lambda x:
                x["confidence"]
        )


    # ========================================================
    # NORMALIZATION USED ONLY FOR CONSENSUS
    # ========================================================

    def consensus_key(self, text):
        """
        Normalize spacing/case ONLY for candidate comparison.

        Actual returned text is never modified.
        """

        text = " ".join(
            text.strip().split()
        )

        return text.casefold()


    # ========================================================
    # CONSENSUS VOTING
    # ========================================================

    def choose_best_candidate(
        self,
        candidates
    ):
        """
        Example:

        full_page   -> INV-100O28  0.78
        upscale2    -> INV-100028  0.94
        upscale3    -> INV-100028  0.97
        clahe       -> INV-100028  0.98
        otsu        -> INV-100028  0.95

        Agreement across preprocessing variants gets
        substantial weight.
        """

        grouped = {}

        for candidate in candidates:

            text = candidate[
                "text"
            ].strip()

            if not text:
                continue

            key = self.consensus_key(
                text
            )

            if key not in grouped:

                grouped[key] = {
                    "texts": [],
                    "confidences": [],
                    "sources": [],
                }

            grouped[key][
                "texts"
            ].append(text)

            grouped[key][
                "confidences"
            ].append(
                float(
                    candidate[
                        "confidence"
                    ]
                )
            )

            grouped[key][
                "sources"
            ].append(
                candidate[
                    "source"
                ]
            )

        ranked = []

        for key, group in grouped.items():

            confidences = group[
                "confidences"
            ]

            support = len(
                confidences
            )

            maximum = max(
                confidences
            )

            average = (
                sum(confidences)
                / support
            )

            # Support bonus capped at 3 agreeing OCR runs.
            support_ratio = min(
                support / 3.0,
                1.0
            )

            # Give meaningful weight to agreement.
            consensus_score = (
                0.55 * average
                +
                0.25 * maximum
                +
                0.20 * support_ratio
            )

            # Pick spelling from the highest-confidence
            # candidate inside this consensus group.
            best_index = int(
                np.argmax(
                    confidences
                )
            )

            best_text = group[
                "texts"
            ][best_index]

            ranked.append({

                "text":
                    best_text,

                "confidence":
                    maximum,

                "average_confidence":
                    average,

                "support":
                    support,

                "sources":
                    group["sources"],

                "consensus_score":
                    consensus_score,
            })

        if not ranked:
            return None

        ranked.sort(
            key=lambda x:
                x["consensus_score"],
            reverse=True,
        )

        return ranked[0]


    # ========================================================
    # AUTOMATIC LOW-CONFIDENCE RETRY
    # ========================================================

    def retry_region(
        self,
        crop,
        original_text,
        original_confidence,
    ):

        variants = self.make_variants(
            crop
        )

        # Initial result participates in voting.
        candidates = [{
            "text":
                original_text,

            "confidence":
                original_confidence,

            "source":
                "full_document",
        }]

        for variant in variants:

            result = (
                self.recognize_variant(
                    variant["image"]
                )
            )

            if result is None:
                continue

            candidates.append({

                "text":
                    result["text"],

                "confidence":
                    result[
                        "confidence"
                    ],

                "source":
                    variant["name"],
            })

        best = self.choose_best_candidate(
            candidates
        )

        return {
            "best": best,
            "all_candidates": candidates,
        }


    # ========================================================
    # READING ORDER
    # ========================================================

    def sort_reading_order(
        self,
        elements
    ):
        """
        Basic invoice/document reading-order reconstruction.

        Groups OCR elements into approximate horizontal rows
        using their vertical centers.
        """

        if not elements:
            return []

        heights = [
            max(
                1,
                item["bbox"][3]
                - item["bbox"][1]
            )
            for item in elements
        ]

        median_height = float(
            np.median(heights)
        )

        y_tolerance = max(
            10,
            median_height * 0.60
        )

        sorted_by_y = sorted(
            elements,
            key=lambda item:
                (
                    (
                        item["bbox"][1]
                        +
                        item["bbox"][3]
                    ) / 2,
                    item["bbox"][0]
                )
        )

        rows = []

        for item in sorted_by_y:

            center_y = (
                item["bbox"][1]
                +
                item["bbox"][3]
            ) / 2

            placed = False

            for row in rows:

                if abs(
                    center_y
                    -
                    row["center_y"]
                ) <= y_tolerance:

                    row[
                        "elements"
                    ].append(item)

                    centers = [
                        (
                            e["bbox"][1]
                            +
                            e["bbox"][3]
                        ) / 2
                        for e
                        in row["elements"]
                    ]

                    row["center_y"] = (
                        sum(centers)
                        / len(centers)
                    )

                    placed = True
                    break

            if not placed:

                rows.append({
                    "center_y":
                        center_y,

                    "elements":
                        [item],
                })

        rows.sort(
            key=lambda row:
                row["center_y"]
        )

        final_order = []

        for row in rows:

            row_elements = sorted(
                row["elements"],
                key=lambda item:
                    item["bbox"][0]
            )

            final_order.extend(
                row_elements
            )

        return final_order


    # ========================================================
    # CREATE PLAIN TEXT WITH ROWS
    # ========================================================

    def build_text_lines(
        self,
        elements
    ):

        if not elements:
            return [], ""

        heights = [
            max(
                1,
                x["bbox"][3]
                - x["bbox"][1]
            )
            for x in elements
        ]

        median_height = float(
            np.median(heights)
        )

        tolerance = max(
            10,
            median_height * 0.60
        )

        rows = []

        for element in elements:

            center_y = (
                element["bbox"][1]
                +
                element["bbox"][3]
            ) / 2

            matched_row = None

            for row in rows:

                if abs(
                    center_y
                    -
                    row["center_y"]
                ) <= tolerance:

                    matched_row = row
                    break

            if matched_row is None:

                matched_row = {
                    "center_y":
                        center_y,

                    "elements":
                        [],
                }

                rows.append(
                    matched_row
                )

            matched_row[
                "elements"
            ].append(element)

        rows.sort(
            key=lambda r:
                r["center_y"]
        )

        text_lines = []

        for row in rows:

            row_elements = sorted(
                row["elements"],
                key=lambda x:
                    x["bbox"][0]
            )

            line = " ".join(
                element["text"]
                for element
                in row_elements
            )

            text_lines.append(
                line
            )

        full_text = "\n".join(
            text_lines
        )

        return text_lines, full_text


    # ========================================================
    # DRAW RESULTS
    # ========================================================

    def draw_results(
        self,
        image,
        elements,
        output_path
    ):

        output = image.copy()

        for element in elements:

            x1, y1, x2, y2 = (
                element["bbox"]
            )

            confidence = element[
                "confidence"
            ]

            # Green-like value not hardcoded by semantic status;
            # simply draw visible rectangles.
            cv2.rectangle(
                output,
                (x1, y1),
                (x2, y2),
                (0, 255, 0),
                2,
            )

            label = (
                f"{element['text']} "
                f"{confidence:.2f}"
            )

            cv2.putText(
                output,
                label,
                (
                    x1,
                    max(
                        y1 - 5,
                        15
                    )
                ),
                cv2.FONT_HERSHEY_SIMPLEX,
                0.45,
                (0, 0, 255),
                1,
                cv2.LINE_AA,
            )

        cv2.imwrite(
            str(output_path),
            output
        )


    # ========================================================
    # MAIN OCR
    # ========================================================

    def extract(
        self,
        image_path,
        output_json=None,
        output_visualization=None,
    ):

        image_path = str(
            image_path
        )

        # ----------------------------------------------------
        # LOAD IMAGE
        # ----------------------------------------------------

        image, metadata = (
            self.prepare_image(
                image_path
            )
        )

        if self.verbose:

            print()
            print(
                "Running full-document OCR..."
            )

        # ----------------------------------------------------
        # FULL DOCUMENT OCR
        # ----------------------------------------------------

        results = self.ocr.predict(

            input=image,

            # Keep all recognized text.
            # We decide what is reliable.
            text_rec_score_thresh=0.0,

            # High recall detection
            text_det_box_thresh=
                self.min_detection_confidence,

            # Keep normal text pixel threshold
            text_det_thresh=0.30,

            return_word_box=True,
        )

        final_elements = []

        processed_image = image

        # ----------------------------------------------------
        # PROCESS PADDLE RESULTS
        # ----------------------------------------------------

        for result in results:

            data = result.json.get(
                "res",
                {}
            )

            texts = list(
                data.get(
                    "rec_texts",
                    []
                )
            )

            scores = list(
                data.get(
                    "rec_scores",
                    []
                )
            )

            boxes = list(
                data.get(
                    "rec_boxes",
                    []
                )
            )

            polygons = list(
                data.get(
                    "rec_polys",
                    []
                )
            )

            detection_scores = list(
                data.get(
                    "dt_scores",
                    []
                )
            )

            # ------------------------------------------------
            # PaddleOCR's OCR coordinates correspond to the
            # preprocessed image if document preprocessing
            # was used.
            # ------------------------------------------------

            try:

                visual_data = (
                    result.img
                )

                if (
                    isinstance(
                        visual_data,
                        dict
                    )
                    and
                    "preprocessed_img"
                    in visual_data
                ):

                    processed_image = (
                        self.to_opencv(
                            visual_data[
                                "preprocessed_img"
                            ]
                        )
                    )

            except Exception:

                processed_image = image

            ph, pw = (
                processed_image.shape[:2]
            )

            # ------------------------------------------------
            # TEXT ELEMENTS
            # ------------------------------------------------

            for index in range(
                min(
                    len(texts),
                    len(scores),
                    len(boxes)
                )
            ):

                text = str(
                    texts[index]
                ).strip()

                if not text:
                    continue

                confidence = float(
                    scores[index]
                )

                box = np.asarray(
                    boxes[index]
                ).reshape(-1)

                if len(box) < 4:
                    continue

                x1 = int(box[0])
                y1 = int(box[1])
                x2 = int(box[2])
                y2 = int(box[3])

                # Clamp coordinates
                x1 = max(
                    0,
                    min(
                        x1,
                        pw - 1
                    )
                )

                y1 = max(
                    0,
                    min(
                        y1,
                        ph - 1
                    )
                )

                x2 = max(
                    x1 + 1,
                    min(
                        x2,
                        pw
                    )
                )

                y2 = max(
                    y1 + 1,
                    min(
                        y2,
                        ph
                    )
                )

                element = {

                    "text":
                        text,

                    "confidence":
                        confidence,

                    "original_text":
                        text,

                    "original_confidence":
                        confidence,

                    "bbox": [
                        x1,
                        y1,
                        x2,
                        y2
                    ],

                    "retried":
                        False,
                }

                # Polygon
                if index < len(
                    polygons
                ):

                    element[
                        "polygon"
                    ] = (
                        np.asarray(
                            polygons[
                                index
                            ]
                        )
                        .astype(int)
                        .tolist()
                    )

                # Detection confidence
                if index < len(
                    detection_scores
                ):

                    try:

                        element[
                            "detection_confidence"
                        ] = float(
                            detection_scores[
                                index
                            ]
                        )

                    except Exception:
                        pass

                # ============================================
                # LOW-CONFIDENCE AUTOMATIC RETRY
                # ============================================

                if (
                    confidence
                    <
                    self.retry_threshold
                ):

                    crop = self.crop_region(
                        processed_image,
                        element["bbox"],
                    )

                    if (
                        crop is not None
                        and
                        crop.size > 0
                    ):

                        retry = (
                            self.retry_region(

                                crop,

                                original_text=
                                    text,

                                original_confidence=
                                    confidence,
                            )
                        )

                        best = retry.get(
                            "best"
                        )

                        if best is not None:

                            element[
                                "retried"
                            ] = True

                            element[
                                "retry_result"
                            ] = retry

                            # --------------------------------
                            # Safer replacement rule
                            #
                            # Don't replace initial OCR merely
                            # because one preprocessing variant
                            # was overconfident.
                            # --------------------------------

                            different = (
                                self.consensus_key(
                                    best["text"]
                                )
                                !=
                                self.consensus_key(
                                    text
                                )
                            )

                            enough_support = (
                                best["support"]
                                >= 2
                            )

                            much_higher_score = (
                                best["confidence"]
                                >=
                                confidence
                                + 0.08
                            )

                            if (
                                different
                                and
                                (
                                    enough_support
                                    or
                                    much_higher_score
                                )
                            ):

                                element[
                                    "text"
                                ] = best[
                                    "text"
                                ]

                                element[
                                    "confidence"
                                ] = best[
                                    "confidence"
                                ]

                                element[
                                    "corrected_by_retry"
                                ] = True

                            else:

                                element[
                                    "corrected_by_retry"
                                ] = False

                final_elements.append(
                    element
                )

        # ----------------------------------------------------
        # READING ORDER
        # ----------------------------------------------------

        final_elements = (
            self.sort_reading_order(
                final_elements
            )
        )

        text_lines, full_text = (
            self.build_text_lines(
                final_elements
            )
        )

        # ----------------------------------------------------
        # RESULT
        # ----------------------------------------------------

        result_data = {

            "image":
                image_path,

            "device":
                self.device,

            "bbox_coordinate_space":
                "preprocessed_image",

            "original_image": {
                "width":
                    metadata[
                        "original_width"
                    ],

                "height":
                    metadata[
                        "original_height"
                    ],
            },

            "ocr_input_image": {
                "width":
                    image.shape[1],

                "height":
                    image.shape[0],

                "scale":
                    metadata[
                        "input_scale"
                    ],
            },

            "processed_image": {
                "width":
                    processed_image.shape[1],

                "height":
                    processed_image.shape[0],
            },

            "text":
                full_text,

            "lines":
                text_lines,

            "elements":
                final_elements,
        }

        # ----------------------------------------------------
        # SAVE JSON
        # ----------------------------------------------------

        if output_json:

            with open(
                output_json,
                "w",
                encoding="utf-8",
            ) as f:

                json.dump(
                    result_data,
                    f,
                    indent=2,
                    ensure_ascii=False,
                )

            if self.verbose:
                print(
                    "JSON saved:",
                    output_json
                )

        # ----------------------------------------------------
        # SAVE BOUNDING BOX IMAGE
        # ----------------------------------------------------

        if output_visualization:

            self.draw_results(
                processed_image,
                final_elements,
                output_visualization,
            )

            if self.verbose:
                print(
                    "Visualization saved:",
                    output_visualization
                )

        # ----------------------------------------------------

        if self.verbose:

            retry_count = sum(
                1
                for x
                in final_elements
                if x.get(
                    "retried",
                    False
                )
            )

            correction_count = sum(
                1
                for x
                in final_elements
                if x.get(
                    "corrected_by_retry",
                    False
                )
            )

            print()
            print("=" * 60)
            print("OCR COMPLETE")
            print(
                "Text regions:",
                len(final_elements)
            )
            print(
                "Retried regions:",
                retry_count
            )
            print(
                "Retry corrections:",
                correction_count
            )
            print("=" * 60)

        return result_data
```

```python
import cv2
import json
import re
import numpy as np
import pytesseract

from pytesseract import Output


class HighAccuracyTesseractOCR:

    def __init__(
        self,
        retry_threshold=80,
        min_confidence=20,
        min_long_side=2000,
        language="eng",
        verbose=True,
    ):
        """
        Tesseract OCR pipeline for document/invoice IMAGES.

        confidence:
            Tesseract confidence is normally 0-100.

        retry_threshold:
            Anything below this confidence gets
            automatically cropped + OCR'd again.

        min_confidence:
            Very poor detections below this are ignored.
        """

        self.retry_threshold = retry_threshold
        self.min_confidence = min_confidence
        self.min_long_side = min_long_side
        self.language = language
        self.verbose = verbose


    # ============================================================
    # LOAD IMAGE
    # ============================================================

    def load_image(self, image_path):

        image = cv2.imread(
            image_path,
            cv2.IMREAD_COLOR
        )

        if image is None:
            raise ValueError(
                f"Could not read image: {image_path}"
            )

        original_h, original_w = image.shape[:2]

        return image, {
            "original_width": original_w,
            "original_height": original_h,
        }


    # ============================================================
    # AUTO ORIENTATION
    # ============================================================

    def correct_orientation(self, image):
        """
        Uses Tesseract OSD to detect page rotation.

        If orientation detection fails,
        original image is returned.
        """

        try:

            rgb = cv2.cvtColor(
                image,
                cv2.COLOR_BGR2RGB
            )

            osd = pytesseract.image_to_osd(
                rgb
            )

            match = re.search(
                r"Rotate: (\d+)",
                osd
            )

            if not match:
                return image, 0

            rotation = int(
                match.group(1)
            )

            if self.verbose:
                print(
                    "Detected rotation:",
                    rotation
                )

            if rotation == 90:

                image = cv2.rotate(
                    image,
                    cv2.ROTATE_90_CLOCKWISE
                )

            elif rotation == 180:

                image = cv2.rotate(
                    image,
                    cv2.ROTATE_180
                )

            elif rotation == 270:

                image = cv2.rotate(
                    image,
                    cv2.ROTATE_90_COUNTERCLOCKWISE
                )

            return image, rotation

        except Exception as e:

            if self.verbose:
                print(
                    "Orientation detection skipped:",
                    str(e)
                )

            return image, 0


    # ============================================================
    # PREPARE FULL IMAGE
    # ============================================================

    def prepare_image(self, image):

        h, w = image.shape[:2]

        long_side = max(
            h,
            w
        )

        scale = 1.0

        if long_side < self.min_long_side:

            scale = (
                self.min_long_side
                / float(long_side)
            )

            image = cv2.resize(
                image,
                None,
                fx=scale,
                fy=scale,
                interpolation=cv2.INTER_CUBIC
            )

        # Very mild preprocessing only.
        #
        # Do NOT threshold the complete invoice here.

        return image, scale


    # ============================================================
    # INITIAL OCR
    # ============================================================

    def full_document_ocr(
        self,
        image
    ):

        rgb = cv2.cvtColor(
            image,
            cv2.COLOR_BGR2RGB
        )

        config = (
            "--oem 3 "
            "--psm 6 "
            "-c preserve_interword_spaces=1"
        )

        data = pytesseract.image_to_data(
            rgb,
            lang=self.language,
            config=config,
            output_type=Output.DICT
        )

        elements = []

        count = len(
            data["text"]
        )

        for i in range(count):

            text = str(
                data["text"][i]
            ).strip()

            if not text:
                continue

            try:

                confidence = float(
                    data["conf"][i]
                )

            except Exception:

                confidence = -1

            if confidence < self.min_confidence:
                continue

            x = int(
                data["left"][i]
            )

            y = int(
                data["top"][i]
            )

            width = int(
                data["width"][i]
            )

            height = int(
                data["height"][i]
            )

            elements.append({

                "text":
                    text,

                "confidence":
                    confidence,

                "original_text":
                    text,

                "original_confidence":
                    confidence,

                "bbox": [
                    x,
                    y,
                    x + width,
                    y + height
                ],

                "block_num":
                    int(
                        data["block_num"][i]
                    ),

                "paragraph_num":
                    int(
                        data["par_num"][i]
                    ),

                "line_num":
                    int(
                        data["line_num"][i]
                    ),

                "word_num":
                    int(
                        data["word_num"][i]
                    ),

                "retried":
                    False,
            })

        return elements


    # ============================================================
    # AUTOMATIC CROP
    # ============================================================

    def crop_region(
        self,
        image,
        bbox
    ):

        x1, y1, x2, y2 = bbox

        h, w = image.shape[:2]

        box_width = max(
            x2 - x1,
            1
        )

        box_height = max(
            y2 - y1,
            1
        )

        # Generous padding
        pad_x = max(
            10,
            int(box_width * 0.20)
        )

        pad_y = max(
            8,
            int(box_height * 0.40)
        )

        x1 = max(
            0,
            x1 - pad_x
        )

        y1 = max(
            0,
            y1 - pad_y
        )

        x2 = min(
            w,
            x2 + pad_x
        )

        y2 = min(
            h,
            y2 + pad_y
        )

        return image[
            y1:y2,
            x1:x2
        ]


    # ============================================================
    # CREATE LOW-CONFIDENCE VARIANTS
    # ============================================================

    def create_variants(
        self,
        crop
    ):

        variants = []

        # --------------------------------------------------------
        # Original crop
        # --------------------------------------------------------

        variants.append({
            "name":
                "original",
            "image":
                crop
        })

        # --------------------------------------------------------
        # 2x upscale
        # --------------------------------------------------------

        up2 = cv2.resize(
            crop,
            None,
            fx=2,
            fy=2,
            interpolation=cv2.INTER_CUBIC
        )

        variants.append({
            "name":
                "upscale_2x",
            "image":
                up2
        })

        # --------------------------------------------------------
        # 3x upscale
        # --------------------------------------------------------

        up3 = cv2.resize(
            crop,
            None,
            fx=3,
            fy=3,
            interpolation=cv2.INTER_CUBIC
        )

        variants.append({
            "name":
                "upscale_3x",
            "image":
                up3
        })

        # --------------------------------------------------------
        # Gray
        # --------------------------------------------------------

        gray = cv2.cvtColor(
            up3,
            cv2.COLOR_BGR2GRAY
        )

        # --------------------------------------------------------
        # CLAHE
        # --------------------------------------------------------

        clahe = cv2.createCLAHE(
            clipLimit=2.0,
            tileGridSize=(8, 8)
        )

        enhanced = clahe.apply(
            gray
        )

        variants.append({
            "name":
                "clahe",

            "image":
                enhanced
        })

        # --------------------------------------------------------
        # Otsu
        # --------------------------------------------------------

        _, otsu = cv2.threshold(
            enhanced,
            0,
            255,
            cv2.THRESH_BINARY
            +
            cv2.THRESH_OTSU
        )

        variants.append({
            "name":
                "otsu",

            "image":
                otsu
        })

        # --------------------------------------------------------
        # Adaptive threshold
        # --------------------------------------------------------

        adaptive = cv2.adaptiveThreshold(
            enhanced,
            255,
            cv2.ADAPTIVE_THRESH_GAUSSIAN_C,
            cv2.THRESH_BINARY,
            31,
            11
        )

        variants.append({
            "name":
                "adaptive",

            "image":
                adaptive
        })

        # --------------------------------------------------------
        # Sharpen
        # --------------------------------------------------------

        sharpen_kernel = np.array([
            [0, -1, 0],
            [-1, 5, -1],
            [0, -1, 0]
        ])

        sharpened = cv2.filter2D(
            enhanced,
            -1,
            sharpen_kernel
        )

        variants.append({
            "name":
                "sharpened",

            "image":
                sharpened
        })

        return variants


    # ============================================================
    # OCR ONE CROP
    # ============================================================

    def recognize_crop(
        self,
        image,
        psm=8
    ):

        config = (
            f"--oem 3 "
            f"--psm {psm}"
        )

        result = pytesseract.image_to_data(

            image,

            lang=self.language,

            config=config,

            output_type=
                Output.DICT
        )

        candidates = []

        for i in range(
            len(result["text"])
        ):

            text = str(
                result["text"][i]
            ).strip()

            if not text:
                continue

            try:

                confidence = float(
                    result["conf"][i]
                )

            except Exception:

                confidence = -1

            if confidence < 0:
                continue

            candidates.append({
                "text":
                    text,

                "confidence":
                    confidence
            })

        if not candidates:
            return None

        return max(
            candidates,
            key=lambda x:
                x["confidence"]
        )


    # ============================================================
    # CONSENSUS
    # ============================================================

    def normalize_text(
        self,
        text
    ):

        return " ".join(
            text.strip().split()
        ).casefold()


    def choose_best_candidate(
        self,
        candidates
    ):

        grouped = {}

        for candidate in candidates:

            text = candidate[
                "text"
            ].strip()

            if not text:
                continue

            key = self.normalize_text(
                text
            )

            if key not in grouped:

                grouped[key] = {
                    "texts": [],
                    "confidences": [],
                    "sources": [],
                }

            grouped[key][
                "texts"
            ].append(
                text
            )

            grouped[key][
                "confidences"
            ].append(
                candidate[
                    "confidence"
                ]
            )

            grouped[key][
                "sources"
            ].append(
                candidate[
                    "source"
                ]
            )

        ranked = []

        for key, group in grouped.items():

            support = len(
                group[
                    "confidences"
                ]
            )

            maximum = max(
                group[
                    "confidences"
                ]
            )

            average = (
                sum(
                    group[
                        "confidences"
                    ]
                )
                / support
            )

            support_ratio = min(
                support / 3,
                1.0
            )

            # Confidence + agreement
            consensus_score = (
                average * 0.55
                +
                maximum * 0.25
                +
                support_ratio * 20
            )

            best_index = int(
                np.argmax(
                    group[
                        "confidences"
                    ]
                )
            )

            ranked.append({

                "text":
                    group[
                        "texts"
                    ][best_index],

                "confidence":
                    maximum,

                "average_confidence":
                    average,

                "support":
                    support,

                "sources":
                    group[
                        "sources"
                    ],

                "consensus_score":
                    consensus_score
            })

        if not ranked:
            return None

        ranked.sort(
            key=lambda x:
                x[
                    "consensus_score"
                ],
            reverse=True
        )

        return ranked[0]


    # ============================================================
    # AUTOMATIC RETRY
    # ============================================================

    def retry_region(
        self,
        crop,
        original_text,
        original_confidence
    ):

        variants = self.create_variants(
            crop
        )

        candidates = [{

            "text":
                original_text,

            "confidence":
                original_confidence,

            "source":
                "initial_document_ocr"
        }]

        for variant in variants:

            # Word-oriented OCR
            result_psm8 = (
                self.recognize_crop(
                    variant["image"],
                    psm=8
                )
            )

            if result_psm8:

                candidates.append({

                    "text":
                        result_psm8[
                            "text"
                        ],

                    "confidence":
                        result_psm8[
                            "confidence"
                        ],

                    "source":
                        variant[
                            "name"
                        ]
                        + "_psm8"
                })

            # Single-line OCR
            result_psm7 = (
                self.recognize_crop(
                    variant["image"],
                    psm=7
                )
            )

            if result_psm7:

                candidates.append({

                    "text":
                        result_psm7[
                            "text"
                        ],

                    "confidence":
                        result_psm7[
                            "confidence"
                        ],

                    "source":
                        variant[
                            "name"
                        ]
                        + "_psm7"
                })

        best = self.choose_best_candidate(
            candidates
        )

        return {

            "best":
                best,

            "all_candidates":
                candidates
        }


    # ============================================================
    # SORT TEXT
    # ============================================================

    def sort_reading_order(
        self,
        elements
    ):

        return sorted(
            elements,
            key=lambda x: (
                x["block_num"],
                x["paragraph_num"],
                x["line_num"],
                x["word_num"]
            )
        )


    # ============================================================
    # RECONSTRUCT LINES
    # ============================================================

    def reconstruct_text(
        self,
        elements
    ):

        lines = {}

        for item in elements:

            key = (
                item[
                    "block_num"
                ],

                item[
                    "paragraph_num"
                ],

                item[
                    "line_num"
                ]
            )

            if key not in lines:
                lines[key] = []

            lines[key].append(
                item
            )

        output_lines = []

        for key in sorted(
            lines.keys()
        ):

            words = sorted(
                lines[key],
                key=lambda x:
                    x["bbox"][0]
            )

            line = " ".join(
                word["text"]
                for word in words
            )

            if line.strip():

                output_lines.append(
                    line
                )

        return (
            output_lines,
            "\n".join(
                output_lines
            )
        )


    # ============================================================
    # DRAW BOUNDING BOXES
    # ============================================================

    def draw_boxes(
        self,
        image,
        elements,
        output_path
    ):

        output = image.copy()

        for item in elements:

            x1, y1, x2, y2 = (
                item["bbox"]
            )

            cv2.rectangle(
                output,
                (x1, y1),
                (x2, y2),
                (0, 255, 0),
                2
            )

            label = (
                f"{item['text']} "
                f"{item['confidence']:.0f}"
            )

            cv2.putText(
                output,
                label,
                (
                    x1,
                    max(
                        15,
                        y1 - 4
                    )
                ),
                cv2.FONT_HERSHEY_SIMPLEX,
                0.4,
                (0, 0, 255),
                1,
                cv2.LINE_AA
            )

        cv2.imwrite(
            output_path,
            output
        )


    # ============================================================
    # MAIN
    # ============================================================

    def extract(
        self,
        image_path,
        output_json=None,
        output_visualization=None
    ):

        # --------------------------------------------------------
        # Read image
        # --------------------------------------------------------

        image, metadata = (
            self.load_image(
                image_path
            )
        )

        # --------------------------------------------------------
        # Orientation
        # --------------------------------------------------------

        image, rotation = (
            self.correct_orientation(
                image
            )
        )

        # --------------------------------------------------------
        # Resize
        # --------------------------------------------------------

        image, scale = (
            self.prepare_image(
                image
            )
        )

        if self.verbose:

            print(
                "OCR image size:",
                image.shape[1],
                "x",
                image.shape[0]
            )

            print(
                "Running initial Tesseract OCR..."
            )

        # --------------------------------------------------------
        # First pass
        # --------------------------------------------------------

        elements = (
            self.full_document_ocr(
                image
            )
        )

        final_elements = []

        # --------------------------------------------------------
        # Automatic retry
        # --------------------------------------------------------

        for element in elements:

            confidence = element[
                "confidence"
            ]

            if (
                confidence
                <
                self.retry_threshold
            ):

                crop = self.crop_region(
                    image,
                    element["bbox"]
                )

                if (
                    crop is not None
                    and
                    crop.size > 0
                ):

                    retry = (
                        self.retry_region(

                            crop,

                            element[
                                "text"
                            ],

                            confidence
                        )
                    )

                    element[
                        "retried"
                    ] = True

                    element[
                        "retry_result"
                    ] = retry

                    best = retry[
                        "best"
                    ]

                    if best:

                        original_normalized = (
                            self.normalize_text(
                                element[
                                    "text"
                                ]
                            )
                        )

                        retry_normalized = (
                            self.normalize_text(
                                best[
                                    "text"
                                ]
                            )
                        )

                        different = (
                            original_normalized
                            !=
                            retry_normalized
                        )

                        enough_support = (
                            best[
                                "support"
                            ]
                            >= 2
                        )

                        confidence_improvement = (
                            best[
                                "confidence"
                            ]
                            >=
                            confidence
                            + 8
                        )

                        if (
                            different
                            and
                            (
                                enough_support
                                or
                                confidence_improvement
                            )
                        ):

                            element[
                                "text"
                            ] = (
                                best[
                                    "text"
                                ]
                            )

                            element[
                                "confidence"
                            ] = (
                                best[
                                    "confidence"
                                ]
                            )

                            element[
                                "corrected_by_retry"
                            ] = True

                        else:

                            element[
                                "corrected_by_retry"
                            ] = False

            final_elements.append(
                element
            )

        # --------------------------------------------------------
        # Reading order
        # --------------------------------------------------------

        final_elements = (
            self.sort_reading_order(
                final_elements
            )
        )

        lines, full_text = (
            self.reconstruct_text(
                final_elements
            )
        )

        # --------------------------------------------------------
        # Result
        # --------------------------------------------------------

        result = {

            "engine":
                "Tesseract",

            "image":
                image_path,

            "rotation":
                rotation,

            "scale":
                scale,

            "original_image": {
                "width":
                    metadata[
                        "original_width"
                    ],

                "height":
                    metadata[
                        "original_height"
                    ]
            },

            "ocr_image": {
                "width":
                    image.shape[1],

                "height":
                    image.shape[0]
            },

            "text":
                full_text,

            "lines":
                lines,

            "elements":
                final_elements
        }

        # --------------------------------------------------------
        # Save JSON
        # --------------------------------------------------------

        if output_json:

            with open(
                output_json,
                "w",
                encoding="utf-8"
            ) as f:

                json.dump(
                    result,
                    f,
                    indent=2,
                    ensure_ascii=False
                )

        # --------------------------------------------------------
        # Save visualization
        # --------------------------------------------------------

        if output_visualization:

            self.draw_boxes(
                image,
                final_elements,
                output_visualization
            )

        # --------------------------------------------------------
        # Summary
        # --------------------------------------------------------

        if self.verbose:

            retried = sum(
                1
                for x in final_elements
                if x.get(
                    "retried",
                    False
                )
            )

            corrected = sum(
                1
                for x in final_elements
                if x.get(
                    "corrected_by_retry",
                    False
                )
            )

            print()
            print("=" * 60)
            print("TESSERACT OCR COMPLETE")
            print("=" * 60)

            print(
                "Words detected:",
                len(final_elements)
            )

            print(
                "Words retried:",
                retried
            )

            print(
                "Words corrected:",
                corrected
            )

        return result
```
